```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler,LabelEncoder,OneHotEncoder
from sklearn.model_selection import train_test_split
import pickle
```

```python
import tensorflow
from tensorflow.keras.models import load_model
```

```python
model=load_model('model.h5')
```

```python
### load one hot encoder and label encoder,standard scaler
with open("one_hot_encoder_geo.pkl",'rb') as file:
    one_hot_encoder_geo=pickle.load(file)
with open("lbl_encoder.pkl",'rb') as file:
    ldl_encoder=pickle.load(file)
with open("scaler.pkl",'rb') as file:
    scaler=pickle.load(file)
```

```python
### take one input
input_data={
    'CreditScore':600,
    'Geography':'France',
    'Gender':'Male',
    'Age':40,
    'Tenure':3,
    'Balance':60000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000
}
```

### convert the Geography column by one hot encoder

```python
encoded_geo=one_hot_encoder_geo.transform([[input_data['Geography']]]).toarray()
encoded_df_geo=pd.DataFrame(encoded_geo,columns=one_hot_encoder_geo.get_feature_names_out(['Geography']))
encoded_df_geo
```

```python
### first convert the input data to dataframe
input_df=pd.DataFrame([input_data])
```

```python
input_df['Gender']=ldl_encoder.transform(input_df['Gender'])
```

```python
### concatenate of two dataframe
input_df=pd.concat([input_df,encoded_df_geo],axis=1)
```

```python
input_df
```

#### drop the geography columns

```python
input_df=input_df.drop('Geography',axis=1)
```

```python
input_df
```

```python
### Apply standard scaler
input_scale=scaler.transform(input_df)
```

```python
input_scale
```

```python
prediction=model.predict(input_scale)
prediction
```

```python
prediction_prob=prediction[0][0]
prediction_prob
```

```python
if prediction_prob>0.5:
    print("customer likely to churn")
else:
    print("customer is not likely to churn")
```
